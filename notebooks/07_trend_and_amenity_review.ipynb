# 07 - Trend and amenity review
The base `plots/` folder contains individual citizen-only charts. `plots_with_trends/` adds per-series descriptive fits. `plots_with_trends_with_averages/` adds selected-cohort averages and compatible residential reference rates. Approved capacity axes use even integers, while advertised capacities and other count axes use integers and retain odd advertised values.

Exact-size fits weight group means by unique parcel counts. Minimum-cutoff fits are descriptive only because their groups overlap. Primary rankings use exact current full/matched cohorts with adequate parcel, event and size-level support; they exclude capacity-selected cohorts and overlapping cutoffs. Group-mean R? is not individual-property explanatory power. Many comparisons were explored, and no causal claim or uncorrected significance claim is made.

Amenity scores use a fixed declared nine-feature physical list, taking the largest score from one listing per parcel. They do not combine separate units into an inflated list. General mentions and event/group language are separate. Poisson association models use robust covariance and control advertised guest size, owner type and license count. Support flags remain visible. These models do not establish causal amenity effects or remove all location confounding.

Review-priority lists rank actual recorded-address calls per parcel. They include listing URLs and evidence, and do not label hosts as proven offenders. Full available dates are retained, with the documented shorter parking-coverage exception. The 59% booking-occupancy normalization is a scenario.
